```python
file_path="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-SkillsNetwork/labs/Data%20files/auto.csv"
```

```python
cars=pd.read_csv(file_path, header=None)
```

```python
cars.info()
cars.describe(include="all")
```

```python
headers = ["symboling","normalized-losses","make","fuel-type","aspiration", "num-of-doors","body-style",
         "drive-wheels","engine-location","wheel-base", "length","width","height","curb-weight","engine-type",
         "num-of-cylinders", "engine-size","fuel-system","bore","stroke","compression-ratio","horsepower",
         "peak-rpm","city-mpg","highway-mpg","price"]
cars.columns=headers
cars.head(20)
```

```python
cars= cars.replace("?", np.nan)
cars
```

```python
missing_values=cars.isnull().sum()
missing_values=(missing_values/len(cars) ) *100
missing_values.sort_values(ascending=False)
```

```python
cars.dtypes
```

```python
cars["normalized-losses"]=cars["normalized-losses"].astype(float)
mean=cars["normalized-losses"].mean()
print(mean)
cars.fillna({"normalized-losses": mean}, inplace=True)
```

```python
missing_values=cars.isnull().sum()
missing_values
```

```python
cars['price']
```

```python
cars.dropna(subset='price', inplace=True, axis=0)
cars
```

```python
missing_values=cars.isnull().sum()
missing_values=(missing_values/len(cars) ) *100
missing_values.sort_values(ascending=False)
```

```python
cars['bore']=cars['bore'].astype(float)
mean=cars['bore'].mean()
cars.fillna({'bore': mean}, inplace=True)
```

```python
missing_values=cars.isnull().sum()
missing_values=(missing_values/len(cars) ) *100
missing_values.sort_values(ascending=False)
```

```python
cars['stroke']=cars['stroke'].astype(float)
mean=cars['stroke'].mean()
cars.fillna({'stroke': mean}, inplace=True)
```

```python
missing_values=cars.isnull().sum()
missing_values=(missing_values/len(cars) ) *100
missing_values.sort_values(ascending=False)
```

```python
cars['num-of-doors']=cars['num-of-doors'].replace(False,np.nan)
freq=cars['num-of-doors'].mode()[0]
print(freq)
cars.fillna({'num-of-doors': freq}, inplace=True)
```

```python
missing_values=cars.isnull().sum()
missing_values=(missing_values/len(cars) ) *100
missing_values.sort_values(ascending=False)
```

```python
cars['horsepower']=cars['horsepower'].astype(float)
mean=cars['horsepower'].mean()
cars.fillna({'horsepower': mean}, inplace=True)
```

```python
missing_values=cars.isnull().sum()
missing_values=(missing_values/len(cars) ) *100
missing_values.sort_values(ascending=False)
```

```python
cars['peak-rpm']=cars['peak-rpm'].astype(float)
mean=cars['peak-rpm'].mean()
cars.fillna({'peak-rpm': mean}, inplace= True)
```

```python
missing_values=cars.isnull().sum()
missing_values=(missing_values/len(cars) ) *100
missing_values.sort_values(ascending=False)
```

```python
cars.reset_index(drop=True, inplace=True)
cars.head(10)
```

```python
cars['normalized-losses']= cars['normalized-losses'].astype(int)
cars['price'] = cars['price'].astype(float)
cars.info()
```

```python
cars.head(5)
```

```python
cars['city-L/100km']=235/cars["city-mpg"]
cars.head(2)
```

```python
cars['highway-L/100km']=235/cars["highway-mpg"]
cars.head(2)
```

```python
cars.info()
```

```python
scaler=MinMaxScaler()
cars['width']= scaler.fit_transform(cars[['width']]) 
cars['width']
```

```python
cars['height']= scaler.fit_transform(cars[['height']]) 
cars['height']
```

```python
cars['length']= scaler.fit_transform(cars[['length']]) 
cars['length']
```

```python
cars['horsepower']=cars['horsepower'].astype(int)
cars['horsepower'].max()
cars['horsepower']
```

```python
plt.pyplot.hist(cars["horsepower"])

plt.pyplot.xlabel("horsepower")
plt.pyplot.ylabel("count")
plt.pyplot.title("horsepower bins")
```

```python
cars['horsepower-binned']=pd.qcut(cars['horsepower'], q=3, labels=['Low', 'Medium', 'High'])
cars[['horsepower','horsepower-binned']].head(20)
```

```python
cars['horsepower-binned'].value_counts()
```

```python
plt.pyplot.bar(['Low', 'Medium', 'High'],cars['horsepower-binned'].value_counts())

plt.pyplot.xlabel("horsepower")
plt.pyplot.ylabel("count")
plt.pyplot.title("hosepower bins")
```

```python
dummies= pd.get_dummies(cars['fuel-type'])
dummies.rename(columns={'gas':'fuel-type-gas', 'diesel':'fuel-type-diesel'}, inplace=True)
```

```python
dummies.tail(5)
```

```python
cars['fuel-type'].tail(5)
```

```python
cars= pd.concat([cars, dummies], axis=1)
```

```python
cars.tail(5)
```

```python
cars.drop(['fuel-type'], axis=1,inplace=True)
```

```python
cars.tail(5)
```

```python
dummies2=pd.get_dummies(cars['aspiration'])
dummies2.rename(columns={'std':'aspiration-std', 'turbo':'aspiration-turbo'}, inplace=True)
dummies2
```

```python
cars=pd.concat([cars, dummies2], axis=1)
cars.drop('aspiration', axis=1, inplace=True)
cars
```

```python
cars.to_csv("clean_df.csv")
```

```python
filepath = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-Coursera/laptop_pricing_dataset_mod1.csv"
laptop = pd.read_csv(filepath)
laptop
```

```python
laptop[['Screen_Size_cm']] = np.round(laptop[['Screen_Size_cm']],2)
laptop.rename(columns={'Unnamed: 0':'index'}, inplace=True)
laptop.replace("?", np.nan)
laptop
```

```python
missing_values=laptop.isnull().sum()
missing_values.sort_values(ascending=False)
```

```python
mean=laptop['Weight_kg'].mean()
laptop.fillna({'Weight_kg': mean}, inplace=True)
```

```python
laptop.isnull().sum().sort_values(ascending=False)
```

```python
freq=laptop['Screen_Size_cm'].mode()[0]
laptop.fillna({'Screen_Size_cm':freq}, inplace=True)
```

```python
missing_values=laptop.isnull().sum()
missing_values.sort_values(ascending=False)
```

```python
laptop
```

```python
laptop['Screen_Size_inch']= laptop['Screen_Size_cm']*2.54
laptop['Weight_pounds']= laptop['Weight_kg'] * 2.205
laptop
```

```python
laptop['CPU_frequency']
```

```python
laptop['CPU_frequency']=scaler.fit_transform(laptop[['CPU_frequency']])
laptop['CPU_frequency']
```

```python
plt.pyplot.hist(laptop['Price'])
plt.pyplot.xlabel("Price")
plt.pyplot.ylabel("number")
plt.pyplot.title("Price graph")
```
