```python
path="baseball.db"
con=sq3.Connection(path)
print(con)
```

```python
query1='''
SELECT * FROM allstarfull;
'''

observations=pds.read_sql(query1, con)
query2='''
SELECT * FROM sqlite_master;
'''

tables=pds.read_sql(query2,con)
```

```python
print(tables)
print(observations)
print(observations.columns.tolist)
```

```python
query3='''
SELECT playerID, SUM(GP) AS game_played, AVG(startingPos) AS avg_starting_position
FROM allstarfull GROUP BY playerID
ORDER BY game_played DESC, avg_starting_position ASC
LIMIT 5
'''

alltime5players=pds.read_sql(query3, con)
print(alltime5players)
```
