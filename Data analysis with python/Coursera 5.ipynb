```python
import pandas as pd
import numpy as np
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
import seaborn as sns
```

```python
filepath = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-SkillsNetwork/labs/Data%20files/automobileEDA.csv"
df = pd.read_csv(filepath)
```

## **SIMPLE LINEAR REGRESSION**

```python
lm=LinearRegression()
```

```python
X = df[['highway-mpg']]
Y = df['price']
```

```python
lm.fit(X, Y)
```

```python
print("le coefficient est : ",lm.coef_, " et la pente est : ", lm.intercept_)
```

```python
Y_predict=lm.predict(X)
r2=r2_score(Y, Y_predict)
mse=mean_squared_error(Y, Y_predict)
print("Le r2 est: ", r2, " et le mse est : ",mse)
```

```python
sns.regplot(x=X, y=Y, data=df)
plt.ylim(0,)
```

```python
ax1=sns.kdeplot(data=Y, color="r", label="Actual Values")
sns.kdeplot(data=Y_predict, color="b", label="Predicted values", ax=ax1)
plt.title("predicted values vs actual")
plt.legend()
plt.show()
```

## **POLYNOMIAL REGRESSION**

```python
pl=PolynomialFeatures(degree=2)
pl
```

```python
W.shape
W_pl=pl.fit_transform(W)
W_pl.shape
```

```python
model=make_pipeline(
    PolynomialFeatures(degree=2),
    StandardScaler(),
    LinearRegression()
)
model
```

```python
model.fit(W, Y)
```

```python
W_predict=model.predict(W)
```

```python
ax1=sns.kdeplot(data=Y, color="r", label="actual Values")
sns.kdeplot(data=W_predict, color="b", label="predicted values", ax=ax1)
plt.legend()
plt.show()
```

```python
r2=r2_score(Y,W_predict)
mse=mean_squared_error(Y, W_predict)
print("Le r2 est: ", r2, " et le mse est : ",mse)
```
